```python
questions = [
    "On average, how many hours do you sleep per night?",
    "How would you rate your overall sleep quality?",
    "How often do you feel sleepy during lectures?",
    "How would you rate your overall classroom productivity?"
]

for question in questions:
    print("\n" + question)
    counts = df[question].value_counts()
    percentages = df[question].value_counts(normalize=True) * 100

    summary = pd.DataFrame({
        "Count": counts,
        "Percentage": percentages.round(2)
    })

    print(summary)
```

```python
# Selecting input features (X)

features = [
    "On average, how many hours do you sleep per night?",
    "5. What time do you usually go to sleep on weekdays?",
    "How consistent is your sleep schedule?",
    "How often do you sleep late because of assignments, studying, projects, etc.?",
    "How would you rate your overall sleep quality?",
    "How often do you feel sleepy during lectures?",
    "14. How much time do you usually spend on your phone/laptop during the hour before sleeping?",
    "How many caffeinated drinks do you usually consume per day?",
    "On average, how many hours do you study outside classroom hours per day?"
]

# Selecting target variable (y)

target = "How would you rate your overall classroom productivity?"

print("Input Features:")
print(features)

print("\nTarget Variable:")
print(target)
```

```python
# Creating input dataset (X)
X = df[features]

# Creating target dataset (y)
y = df[target]

# Displaying the first 5 records
print("Input Features (X):")
display(X.head())

print("\nTarget Variable (y):")
display(y.head())

# Checking dimensions
print("\nShape of X:", X.shape)
print("Shape of y:", y.shape)
```

```python
# Checking data types of input features
print("Data types of input features:")
print(X.dtypes)

# Checking target variable distribution
print("\nClassroom Productivity Distribution:")
print(y.value_counts().sort_index())

# Checking unique values in target
print("\nUnique productivity ratings:")
print(y.unique())
```

```python
# Inspecting categorical feature values

categorical_columns = X.select_dtypes(include=['object']).columns

for column in categorical_columns:
    print("\nColumn:", column)
    print(X[column].value_counts(dropna=False))
```

```python
from sklearn.model_selection import train_test_split

# Splitting the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Displaying dataset sizes
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Identifying categorical and numerical columns
categorical_columns = X_train.select_dtypes(include=['object']).columns
numerical_columns = X_train.select_dtypes(include=['number']).columns

print("Categorical columns:", list(categorical_columns))
print("\nNumerical columns:", list(numerical_columns))
```

```python
# Creating the preprocessing transformer

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_columns),
        ('num', 'passthrough', numerical_columns)
    ]
)

# Encoding training and testing data
X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Training data after encoding:", X_train_encoded.shape)
print("Testing data after encoding:", X_test_encoded.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

# Creating the Decision Tree model
model = DecisionTreeClassifier(
    random_state=42,
    max_depth=4
)

print("Decision Tree model created successfully!")
```

```python
# Training the model
model.fit(X_train_encoded, y_train)

print("Model training completed successfully!")
```

```python
# Predicting classroom productivity using test data

y_pred = model.predict(X_test_encoded)

print("Actual Productivity Ratings:")
print(y_test.values)

print("\nPredicted Productivity Ratings:")
print(y_pred)
```

```python
# Comparing actual and predicted values

comparison = pd.DataFrame({
    "Actual Productivity": y_test.values,
    "Predicted Productivity": y_pred
})

display(comparison)
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Calculating model accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

# Generating classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred, labels=[1, 2, 3, 4, 5], zero_division=0))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Creating confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=[1, 2, 3, 4, 5])

# Displaying confusion matrix
display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[1, 2, 3, 4, 5]
)

display.plot(cmap='Blues')

plt.title("Confusion Matrix - Decision Tree Classifier")
plt.show()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import StratifiedKFold, cross_validate

# Creating different ML models

models = {
    "Dummy Classifier": DummyClassifier(strategy="most_frequent"),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=4,
        random_state=42,
        class_weight="balanced"
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        class_weight="balanced"
    )
}

# Cross-validation setup
cv = StratifiedKFold(
    n_splits=4,
    shuffle=True,
    random_state=42
)

# Comparing model performance
for name, classifier in models.items():

    pipeline = Pipeline([
        ("preprocessing", preprocessor),
        ("classifier", classifier)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=["accuracy", "f1_macro", "f1_weighted"]
    )

    print("\nModel:", name)
    print("Average Accuracy:",
          round(scores["test_accuracy"].mean() * 100, 2), "%")

    print("Macro F1-score:",
          round(scores["test_f1_macro"].mean(), 3))

    print("Weighted F1-score:",
          round(scores["test_f1_weighted"].mean(), 3))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

# Creating Random Forest pipeline
rf_model = Pipeline([
    ("preprocessing", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        class_weight="balanced"
    ))
])

# Training the model
rf_model.fit(X_train, y_train)

print("Random Forest training completed!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

# Predicting classroom productivity
rf_predictions = rf_model.predict(X_test)

# Calculating accuracy
rf_accuracy = accuracy_score(y_test, rf_predictions)

print("Random Forest Test Accuracy:",
      round(rf_accuracy * 100, 2), "%")

# Classification report
print("\nClassification Report:")

print(classification_report(
    y_test,
    rf_predictions,
    labels=[1, 2, 3, 4, 5],
    zero_division=0
))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_predictions(
    y_test,
    rf_predictions,
    labels=[1, 2, 3, 4, 5],
    cmap="Greens"
)

plt.title("Confusion Matrix - Random Forest")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_validate, StratifiedKFold

# Creating a new preprocessing pipeline
scaled_preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_columns),
        ('num', StandardScaler(), numerical_columns)
    ]
)

# Testing different K values
for k in [3, 5, 7, 9, 11]:

    knn_pipeline = Pipeline([
        ("preprocessing", scaled_preprocessor),
        ("classifier", KNeighborsClassifier(n_neighbors=k))
    ])

    scores = cross_validate(
        knn_pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=["accuracy", "f1_macro"]
    )

    print(
        "K =", k,
        "| Accuracy =", round(scores["test_accuracy"].mean() * 100, 2), "%",
        "| Macro F1 =", round(scores["test_f1_macro"].mean(), 3)
    )
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Fresh preprocessing for cross-validation
preprocessor_cv = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_columns),
    ("num", "passthrough", numerical_columns)
])

# Models to compare
models = {
    "Dummy Classifier": DummyClassifier(strategy="most_frequent"),
    "Decision Tree": DecisionTreeClassifier(
        random_state=42, max_depth=4, class_weight="balanced"
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=100, random_state=42, class_weight="balanced"
    ),
    "Logistic Regression": LogisticRegression(
        max_iter=2000, class_weight="balanced"
    ),
    "SVM": SVC(class_weight="balanced"),
    "KNN (K=3)": KNeighborsClassifier(n_neighbors=3)
}

cv = StratifiedKFold(n_splits=4, shuffle=True, random_state=42)

for name, classifier in models.items():

    pipeline = Pipeline([
        ("preprocessing", preprocessor_cv),
        ("classifier", classifier)
    ])

    scores = cross_validate(
        pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring=["accuracy", "f1_macro", "f1_weighted"]
    )

    print(
        name,
        "| Accuracy:", round(scores["test_accuracy"].mean() * 100, 2), "%",
        "| Macro F1:", round(scores["test_f1_macro"].mean(), 3),
        "| Weighted F1:", round(scores["test_f1_weighted"].mean(), 3)
    )
```

```python
# Creating selected KNN model
final_model = Pipeline([
    ("preprocessing", preprocessor_cv),
    ("classifier", KNeighborsClassifier(n_neighbors=3))
])

# Training on the complete training dataset
final_model.fit(X_train, y_train)

print("Final KNN model trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Generate predictions
final_predictions = final_model.predict(X_test)

# Calculate accuracy
print("Final Model Accuracy:",
      round(accuracy_score(y_test, final_predictions) * 100, 2), "%")

# Classification report
print("\nClassification Report:")
print(classification_report(y_test, final_predictions, zero_division=0))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, final_predictions))
```
